## Total cost of a lockdown function

```python
def totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1End, lockdown2Start, lockdown2End, W):
    
    totalCost = 0
    
    for i in range(len(efficaciesArray)):
        adultsVaccine = outbreakFunctions.vaccinationDay(elderlyAgeGrp, vaccineStart, efficaciesArray[i], lockdown1Start, lockdown1End, lockdown2Start, lockdown2End, 0.4)
        childrenVaccine = outbreakFunctions.vaccinationDay(adultAgeGrp, adultsVaccine, efficaciesArray[i], lockdown1Start, lockdown1End, lockdown2Start, lockdown2End, 0.4)
        
        elderlySim = outbreakFunctions.SEIHRVDsim_2Variants(elderlyAgeGrp, vaccineStart, efficaciesArray[i], lockdown1Start, lockdown1End, lockdown2Start, lockdown2End)
        adultSim = outbreakFunctions.SEIHRVDsim_2Variants(adultAgeGrp, adultsVaccine, efficaciesArray[i], lockdown1Start, lockdown1End, lockdown2Start, lockdown2End)
        youngSim = outbreakFunctions.SEIHRVDsim_2Variants(youngAgeGrp, childrenVaccine, efficaciesArray[i], lockdown1Start, lockdown1End, lockdown2Start, lockdown2End)

        hospY, deadY, vaccY = youngSim.outbreakEndResults()
        hospA, deadA, vaccA = adultSim.outbreakEndResults()
        hospE, deadE, vaccE = elderlySim.outbreakEndResults()

        fatalQALY = (23.5*deadY + 17.6*deadA + 5.3*deadE)
        hospQALY = 10/365 * (0.948*hospY + 0.881*hospA + 0.738*hospE)

        monthlyGDPnoLD = 103.6397
        monthlyGDPinLD = 103.6397 - 50.2415*0.8 + 116.9759*(0.8**2) - 90.7837*(0.8**4)

        dailyGDPloss = (monthlyGDPnoLD*(10**9) - monthlyGDPinLD*(10**9))/30

        netMonetaryLoss = W*(fatalQALY + hospQALY) + dailyGDPloss*((lockdown1End - lockdown1Start) + (lockdown2End - lockdown2Start))
        
        totalCost += weights[i] * netMonetaryLoss
        
    return(totalCost)
```

## Array of costs for different lockdown lengths (for plotting)

```python
def costsOverLockdowns(efficaciesArray, weights, vaccineStart, lockdown1Start, wks):
    costArray1_LD1 = []
    costArray2_LD1 = []
    costArray3_LD1 = []
    
    for i in wks:
        cost1 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+i*7, 0, 0, 20000)
        cost2 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+i*7, 0, 0, 50000)
        cost3 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+i*7, 0, 0, 100000)
        
        costArray1_LD1.append(cost1)
        costArray2_LD1.append(cost2)
        costArray3_LD1.append(cost3)
        
    costArrayLD1 = [costArray1_LD1, costArray2_LD1, costArray3_LD1]

    # Adding a second 4-week lockdown 20 days after LD1 ends   
    costArray1_LD2_20D = []
    costArray2_LD2_20D = []
    costArray3_LD2_20D = []
    for j in wks:
        cost1 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 20, lockdown1Start+j*7 + 48, 20000)
        cost2 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 20, lockdown1Start+j*7 + 48, 50000)
        cost3 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 20, lockdown1Start+j*7 + 48, 100000)
        
        costArray1_LD2_20D.append(cost1)
        costArray2_LD2_20D.append(cost2)
        costArray3_LD2_20D.append(cost3)

    costArrayLD2_20D = [costArray1_LD2_20D, costArray2_LD2_20D, costArray3_LD2_20D]
    
    # Adding a second 4-week lockdown 40 days after LD1 ends 
    costArray1_LD2_40D = []
    costArray2_LD2_40D = []
    costArray3_LD2_40D = []
    for j in wks:
        cost1 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 40, lockdown1Start+j*7 + 68, 20000)
        cost2 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 40, lockdown1Start+j*7 + 68, 50000)
        cost3 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 40, lockdown1Start+j*7 + 68, 100000)
        
        costArray1_LD2_40D.append(cost1)
        costArray2_LD2_40D.append(cost2)
        costArray3_LD2_40D.append(cost3)
        
    costArrayLD2_40D = [costArray1_LD2_40D, costArray2_LD2_40D, costArray3_LD2_40D]
    
    # Adding a second 4-week lockdown 60 days after LD1 ends 
    costArray1_LD2_60D = []
    costArray2_LD2_60D = []
    costArray3_LD2_60D = []
    for j in wks:
        cost1 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 60, lockdown1Start+j*7 + 88, 20000)
        cost2 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 60, lockdown1Start+j*7 + 88, 50000)
        cost3 = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+j*7, lockdown1Start+j*7 + 60, lockdown1Start+j*7 + 88, 100000)
        
        costArray1_LD2_60D.append(cost1)
        costArray2_LD2_60D.append(cost2)
        costArray3_LD2_60D.append(cost3)
        
    costArrayLD2_60D = [costArray1_LD2_60D, costArray2_LD2_60D, costArray3_LD2_60D]
    
    return(costArrayLD1, costArrayLD2_20D, costArrayLD2_40D, costArrayLD2_60D)
```

## Optimal weeks of lockdown for a given vaccine weighting

```python
def optimalStrategy(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown2, W, wks):
    weekIncrement = wks[1]-wks[0]
    
    costArray = []
    
    if lockdown2 == True: #If there is a second lockdown:
        for i in wks:
            cost = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+i*7, lockdown1Start+i*7 + 40, lockdown1Start+i*7 + 68, W)
            
            costArray.append(cost)
    else: #No second lockdown:
        for i in wks:  
            cost = totalCost(efficaciesArray, weights, vaccineStart, lockdown1Start, lockdown1Start+i*7, 0, 0, W)
            
            costArray.append(cost)
    
    minimisedCost = min(costArray)
    optimalLockdown = costArray.index(minimisedCost) * weekIncrement
    
    return(optimalLockdown)
```

## Discretising optimal LD weeks and creating an array of all optimal solutions for all weights (for plotting)

```python
def get_category(weeks):
    if weeks == 0:
        return 0
    if 1 <= weeks <= 5:
        return 1
    if 6 <= weeks <= 10:
        return 2
    if 11 <= weeks <= 15:
        return 3
    if 16 <= weeks <= 20:
        return 4
    else:
        return 5


def optimalLockdownCategory(efficaciesArray, vaccineStart, lockdown1Start, lockdown2, W, wks, n):
    n = 10  # n=10 gives weights 0, 0.1, 0.2, 0.3,...
    w1, w2, w3 = [], [], []
    results = []

    for i in range(n + 1):
        for j in range(n + 1 - i):
            k = n - i - j

            v1, v2, v3 = i/n, j/n, k/n
            w1.append(v1)
            w2.append(v2)
            w3.append(v3)

            optimalLockdown = optimalStrategy(efficaciesArray, [v1,v2,v3], vaccineStart, lockdown1Start, False, W, wks)
            
            results.append(get_category(optimalLockdown))
            
    return(w1,w2,w3, results)
```

```python
efficaciesArray = [0.5, 0.7, 0.9]
vaccineStart = 70
lockdown1Start = 40
lockdown2 = False  #T or F depending on if there is a second lockdown
n = 10
weekStart, weekEnd, weekIncrement = 0, 30, 2 
wks = np.arange(weekStart, weekEnd, weekIncrement)

w11, w21, w31, resultsW1 = optimalLockdownCategory(efficaciesArray, vaccineStart, lockdown1Start, lockdown2, 20000, wks, n)
```

```python
colors = ['red', 'orange', 'yellow', 'green', 'blue', 'pink']
cmap = mcolors.ListedColormap(colors)
bounds = [0, 1, 2, 3, 4, 5, 6] 
norm = mcolors.BoundaryNorm(bounds, cmap.N)

fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(projection='ternary')

plot = ax.tripcolor(w11, w21, w31, resultsW1, cmap=cmap, norm=norm, shading='flat')
ax.set_tlabel('$w_1$', fontsize=20)
ax.set_llabel('$w_2$', fontsize=20)
ax.set_rlabel('$w_3$', fontsize=20)
ax.taxis.set_label_position('tick1')
ax.laxis.set_label_position('tick1')
ax.raxis.set_label_position('tick1')


cax = ax.inset_axes([1.05, 0.1, 0.05, 0.9], transform=ax.transAxes)
colorbar = fig.colorbar(plot, cax=cax, ticks=[0.5,1.5,2.5,3.5,4.5, 5.5])
colorbar.ax.set_yticklabels(['None', '1-5 weeks', '6-10 weeks', '11-15 weeks', '16-20 weeks', '21+ weeks'])
colorbar.set_label("Weeks of Lockdown", rotation=270, fontsize = 15, va="baseline")


plt.show()
```
